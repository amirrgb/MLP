```python
X_train, X_test, y_train, y_test, train_loader, test_loader = get_moons_data()

plot_data(X_train,y_train,"Moons Dataset (train set)")
```

```python
neuron_count = [1,2,5,20]
criterion = ["BCEWithLogitsLoss", "BCELoss"]
optimizer = ["SGD", "Adam"]

loss_histories = {}
acc_histories = {}

for nc in neuron_count:
    result = run_experiment(
        model_class=MLP,
        train_loader=train_loader,
        test_loader=test_loader,
        lr=0.1,
        epochs=1000,
        optimizer_name=optimizer[0],
        criterion_name=criterion[0],
    	dataset_name=f"moons_MLP_hidden_{nc}",
        plot_decision=True,
        X_vis=X_train,
        y_vis=y_train,
        model_kwargs={"hidden_dim": nc}
    )

    loss_histories[f"train neurons={nc}"] = result["train_loss"]
    loss_histories[f"test neurons={nc}"] = result["test_loss"]
    acc_histories[f"train neurons={nc}"] = result["train_acc"]
    acc_histories[f"test neurons={nc}"] = result["test_acc"]
```

```python
plot_multi_training_curves(
    loss_histories,
    "MLP - moons - Neuron Count Comparison (Train & Test Loss)",
    "mlp_nc_comparison"
)    

plot_multi_accuracy_curves(
    acc_histories,
    "MLP - moons - Neuron Count Comparison (Train & Test acc)",
    "mlp_nc_comparison"
)    
```
